```python
import numpy as np
import matplotlib.pyplot as plt
from skimage.color import rgb2gray
from skimage import data
from skimage.filters import gaussian
from skimage.segmentation import active_contour
```

```python
# img = data.astronaut()
img = iio.imread(uri='test/images/test3.jpg')
img = rgb2gray(img)

s = np.linspace(0, 2 * np.pi, 400)
r = 200 + 100 * np.sin(s)
c = 400 + 100 * np.cos(s)
init = np.array([r, c]).T

snake = active_contour(
    gaussian(img, sigma=3, preserve_range=False),
    init,
    alpha=0.015,
    beta=10,
    gamma=0.001,
)

fig, ax = plt.subplots(figsize=(7, 7))
ax.imshow(img, cmap=plt.cm.gray)
ax.plot(init[:, 1], init[:, 0], '--r', lw=3)
ax.plot(snake[:, 1], snake[:, 0], '-b', lw=3)
ax.set_xticks([]), ax.set_yticks([])
ax.axis([0, img.shape[1], img.shape[0], 0])

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

from scipy.ndimage import gaussian_filter
from skimage import data
from skimage import img_as_float
from skimage.morphology import reconstruction
img = iio.imread(uri='test/images/test2.webp')

# Convert to float: Important for subtraction later which won't work with uint8
image = img_as_float(img)
image = gaussian_filter(image, 1)

seed = np.copy(image)
seed[1:-1, 1:-1] = image.min()
mask = image

dilated = reconstruction(seed, mask, method='dilation')
```

```python
fig, (ax0, ax1, ax2) = plt.subplots(
    nrows=1, ncols=3, figsize=(8, 2.5), sharex=True, sharey=True
)

ax0.imshow(image, cmap='gray')
ax0.set_title('original image')
ax0.axis('off')

ax1.imshow(dilated, vmin=image.min(), vmax=image.max(), cmap='gray')
ax1.set_title('dilated')
ax1.axis('off')

ax2.imshow(image - dilated, cmap='gray')
ax2.set_title('image - dilated')
ax2.axis('off')

fig.tight_layout()
```

```python
import matplotlib
import matplotlib.pyplot as plt
import numpy as np

from skimage import data, img_as_float
from skimage import exposure


matplotlib.rcParams['font.size'] = 8


def plot_img_and_hist(image, axes, bins=256):
    """Plot an image along with its histogram and cumulative histogram."""
    image = img_as_float(image)
    ax_img, ax_hist = axes
    ax_cdf = ax_hist.twinx()

    # Display image
    ax_img.imshow(image, cmap=plt.cm.gray)
    ax_img.set_axis_off()

    # Display histogram
    ax_hist.hist(image.ravel(), bins=bins, histtype='step', color='black')
    ax_hist.ticklabel_format(axis='y', style='scientific', scilimits=(0, 0))
    ax_hist.set_xlabel('Pixel intensity')
    ax_hist.set_xlim(0, 1)
    ax_hist.set_yticks([])

    # Display cumulative distribution
    img_cdf, bins = exposure.cumulative_distribution(image, bins)
    ax_cdf.plot(bins, img_cdf, 'r')
    ax_cdf.set_yticks([])

    return ax_img, ax_hist, ax_cdf


# Load an example image
# img = data.moon()
img = iio.imread(uri='test/images/test2.webp')

# Contrast stretching
p2, p98 = np.percentile(img, (2, 98))
img_rescale = exposure.rescale_intensity(img, in_range=(p2, p98))

# Equalization
img_eq = exposure.equalize_hist(img)

# Adaptive Equalization
img_adapteq = exposure.equalize_adapthist(img, clip_limit=0.03)

# Display results
fig = plt.figure(figsize=(8, 5))
axes = np.zeros((2, 4), dtype=object)
axes[0, 0] = fig.add_subplot(2, 4, 1)
for i in range(1, 4):
    axes[0, i] = fig.add_subplot(2, 4, 1 + i, sharex=axes[0, 0], sharey=axes[0, 0])
for i in range(0, 4):
    axes[1, i] = fig.add_subplot(2, 4, 5 + i)

ax_img, ax_hist, ax_cdf = plot_img_and_hist(img, axes[:, 0])
ax_img.set_title('Low contrast image')

y_min, y_max = ax_hist.get_ylim()
ax_hist.set_ylabel('Number of pixels')
ax_hist.set_yticks(np.linspace(0, y_max, 5))

ax_img, ax_hist, ax_cdf = plot_img_and_hist(img_rescale, axes[:, 1])
ax_img.set_title('Contrast stretching')

ax_img, ax_hist, ax_cdf = plot_img_and_hist(img_eq, axes[:, 2])
ax_img.set_title('Histogram equalization')

ax_img, ax_hist, ax_cdf = plot_img_and_hist(img_adapteq, axes[:, 3])
ax_img.set_title('Adaptive equalization')

ax_cdf.set_ylabel('Fraction of total intensity')
ax_cdf.set_yticks(np.linspace(0, 1, 5))

# prevent overlap of y-axis labels
fig.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.widgets import Slider
from scipy import ndimage as ndi
from skimage.util import random_noise
from skimage import feature
from skimage.color import rgb2gray
import imageio.v3 as iio
```

```python

# Load image
img = iio.imread('test/images/test3.jpg')
image = rgb2gray(img)

# Initial sigma
initial_sigma = 1.0

# Compute initial Canny edge detection
edges = feature.canny(image, sigma=initial_sigma)

# Create plot
fig, ax = plt.subplots(nrows=1, ncols=2, figsize=(10, 4))
plt.subplots_adjust(bottom=0.25)

# Original image
ax[0].imshow(image, cmap='gray')
ax[0].set_title('Original image')
ax[0].axis('off')

# Edge-detected image
edge_plot = ax[1].imshow(edges, cmap='gray')
ax[1].set_title(f'Canny edges (σ={initial_sigma})')
ax[1].axis('off')

# Slider setup
ax_slider = plt.axes([0.25, 0.1, 0.5, 0.03])
sigma_slider = Slider(ax_slider, 'Sigma', 0.1, 5.0, valinit=initial_sigma, valstep=0.1)

# Update function
def update(val):
    sigma = sigma_slider.val
    updated_edges = feature.canny(image, sigma=sigma)
    edge_plot.set_data(updated_edges)
    ax[1].set_title(f'Canny edges (σ={sigma:.1f})')
    fig.canvas.draw_idle()

sigma_slider.on_changed(update)

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider
from skimage import feature
from skimage.color import rgb2gray
import imageio.v3 as iio

# Load and preprocess image
img = iio.imread('test/images/test3.jpg')
gray = rgb2gray(img)

# Define interactive function
def update_canny(sigma):
    edges = feature.canny(gray, sigma=sigma)
    
    plt.figure(figsize=(10, 4))
    
    plt.subplot(1, 2, 1)
    plt.imshow(gray, cmap='gray')
    plt.title("Original Image")
    plt.axis('off')
    
    plt.subplot(1, 2, 2)
    plt.imshow(edges, cmap='gray')
    plt.title(f"Canny Edges (σ={sigma:.2f})")
    plt.axis('off')
    
    plt.tight_layout()
    plt.show()

# Display interactive slider
interact(update_canny, sigma=FloatSlider(value=1.0, min=0.1, max=5.0, step=0.1))
```

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import FloatSlider, Button, VBox, HBox, Output
from skimage import feature
from skimage.color import rgb2gray
 

# Load and preprocess image
img = iio.imread('test/images/test3.jpg')
gray = rgb2gray(img)

# Create widgets
sigma_slider = FloatSlider(value=1.0, min=0.1, max=5.0, step=0.1, description='Sigma:')
update_button = Button(description='Update')
output = Output()

# Define button click handler
def on_button_click(b):
    with output:
        output.clear_output(wait=True)
        sigma = sigma_slider.value
        edges = feature.canny(gray, sigma=sigma)
        
        plt.figure(figsize=(10, 4))
        plt.subplot(1, 2, 1)
        plt.imshow(gray, cmap='gray')
        plt.title("Original Image")
        plt.axis('off')
        
        plt.subplot(1, 2, 2)
        plt.imshow(edges, cmap='gray')
        plt.title(f"Canny Edges (σ={sigma:.2f})")
        plt.axis('off')
        
        plt.tight_layout()
        plt.show()

# Connect button to function
update_button.on_click(on_button_click)

# Display UI
display(VBox([HBox([sigma_slider, update_button]), output]))
```

```python
# Importing the required libraries
import skimage.io as io
import skimage
import numpy as np
import matplotlib.pyplot as plt
from skimage.color import rgb2gray
from skimage.filters import gaussian 
from skimage.segmentation import active_contour
from skimage.segmentation import random_walker
from skimage import draw
from skimage.segmentation import slic, mark_boundaries
from skimage.color import label2rgb
from skimage.segmentation import felzenszwalb, mark_boundaries
```

```python
#Load the image
lady = io.imread('test/images/test3.jpg')
#Turn the original image to grayscale
gray_lady = rgb2gray(lady)
#Denoise the image
image = gaussian(gray_lady, 1.5, preserve_range=False)
#We will define a circle that encircles the area of interest:
#Create 400 points (in radians) in between 0 and 2*Pi
#By using these points draw a circle with the help of
# sine and cosine functions. The radius is 220 and the 
# center positioned right in the middle of the area of interest.
s = np.linspace(0, 2*np.pi, 400)
r = 500 + 220*np.sin(s)
c = 700 + 220*np.cos(s)
#Create the (x,y) coordinates of the circle by combining 
#two lists of data points
init = np.array([r,c]).T
#Apply Active Contour algorithm to the specified area:
snake = active_contour(image,
init, alpha=0.0013, beta=10, gamma=0.0007)
#Draw the pre-defined area and created contour around 
#the face on the image:
fig, ax = plt.subplots(figsize=(7, 7))
ax.set_title('Object Detection with Active Contour (Snake)')
ax.imshow(gray_lady, cmap=plt.cm.gray)
ax.plot(init[:, 1], init[:, 0], '--r', lw=3)
ax.plot(snake[:, 1], snake[:, 0], '-b', lw=3)
ax.set_xticks([]), ax.set_yticks([])
ax.legend(['Area of Interest', 'Detected Object'])
ax.axis([0, image.shape[1], image.shape[0], 0])
plt.show()
```

```python
from rembg import remove
from PIL import Image
import matplotlib.pyplot as plt
import numpy as np
from skimage import measure
import io

# Load input image and remove background
input_path = 'test/images/test3.jpg'  # replace with your image path

img = Image.open(input_path).convert('RGBA')

img_bytes = io.BytesIO()
img.save(img_bytes, format='PNG')
img_bytes = img_bytes.getvalue()

output_bytes = remove(img_bytes)
output_img = Image.open(io.BytesIO(output_bytes)).convert('RGBA')

# --- STEP 2: Threshold the alpha channel ---
alpha = np.array(output_img.split()[-1])  # extract alpha channel
binary_mask = alpha > 0  # threshold: foreground = True, background = False

# --- STEP 3: Extract contours ---
# Find contours at a constant value of 0.5 (True/False boundary)
contours = measure.find_contours(binary_mask, level=0.5)

# --- Visualization ---
fig, ax = plt.subplots(1, 3, figsize=(18, 6))

ax[0].imshow(img)
ax[0].set_title('Original Image')
ax[0].axis('off')

ax[1].imshow(binary_mask, cmap='gray')
ax[1].set_title('Thresholded Alpha Mask')
ax[1].axis('off')

ax[2].imshow(binary_mask, cmap='gray')
for contour in contours:
    ax[2].plot(contour[:, 1], contour[:, 0], linewidth=2, color='red')
ax[2].set_title('Vector Outlines (Contours)')
ax[2].axis('off')

plt.tight_layout()
plt.show()
```

```python
from rembg import remove
from PIL import Image
import matplotlib.pyplot as plt
import numpy as np
from skimage import measure
from scipy.spatial import ConvexHull
import io

# Step 1: Load & remove background
input_path = 'test/images/test4.jpg'
img = Image.open(input_path).convert('RGBA')

img_bytes = io.BytesIO()
img.save(img_bytes, format='PNG')
img_bytes = img_bytes.getvalue()

output_bytes = remove(img_bytes)
output_img = Image.open(io.BytesIO(output_bytes)).convert('RGBA')

# Step 2: Extract alpha mask and threshold
alpha = np.array(output_img.split()[-1])
binary_mask = alpha > 0

# Step 3: Extract contours
contours = measure.find_contours(binary_mask, level=0.5)

# Step 4: Merge all contours into one point cloud
all_points = np.vstack(contours)

# Step 5: Compute convex hull
hull = ConvexHull(all_points)

# Step 6: Visualization
fig, ax = plt.subplots(1, 2, figsize=(14, 6))

# Show binary mask + contours
ax[0].imshow(binary_mask, cmap='gray')
for contour in contours:
    ax[0].plot(contour[:, 1], contour[:, 0], linewidth=1, color='white')
ax[0].set_title('Original Contours')
ax[0].axis('off')

# Show convex hull
ax[1].imshow(binary_mask, cmap='gray')
# Close the polygon by repeating the first point
hull_points = all_points[hull.vertices]
hull_points = np.vstack([hull_points, hull_points[0]])
ax[1].plot(hull_points[:, 1], hull_points[:, 0], 'r-', linewidth=2)
ax[1].set_title('Convex Hull Outline')
ax[1].axis('off')

plt.tight_layout()
plt.show()
```

```python
from rembg import remove
from PIL import Image
import matplotlib.pyplot as plt
import numpy as np
from skimage import measure
from scipy.spatial import ConvexHull
from scipy.ndimage import label, find_objects
import io
from scipy.interpolate import splprep, splev

# Convert contour to a spline-smoothed path
def smooth_contour(points, smoothing=0.01, num_points=200):
    x, y = points[:, 0], points[:, 1]
    tck, _ = splprep([x, y], s=smoothing)
    u_fine = np.linspace(0, 1, num_points)
    x_smooth, y_smooth = splev(u_fine, tck)
    return np.vstack([x_smooth, y_smooth]).T

# Step 1: Load & remove background
input_path = 'test/images/test6.jpg'
img = Image.open(input_path).convert('RGBA')

img_bytes = io.BytesIO()
img.save(img_bytes, format='PNG')
img_bytes = img_bytes.getvalue()

output_bytes = remove(img_bytes)
output_img = Image.open(io.BytesIO(output_bytes)).convert('RGBA')

# Step 2: Threshold the alpha channel
alpha = np.array(output_img.split()[-1])
binary_mask = alpha > 0

# Step 3: Label connected components
labeled, num_objects = label(binary_mask)
slices = find_objects(labeled)

# Step 4: Convex hulls for each object
fig, ax = plt.subplots(figsize=(8, 8))
ax.imshow(binary_mask, cmap='gray')

for obj_id in range(1, num_objects + 1):
    mask = labeled == obj_id
    contours = measure.find_contours(mask, level=0.5)
    
    if not contours:
        continue

    # Merge all contours of this object
    points = np.vstack(contours)
    
    # Compute convex hull
    if len(points) >= 3:
        hull = ConvexHull(points)
        hull_pts = points[hull.vertices]
        hull_pts = np.vstack([hull_pts, hull_pts[0]])  # close the loop
        hull_pts = smooth_contour(hull_pts, smoothing=5.0)
        ax.plot(hull_pts[:, 1], hull_pts[:, 0], color='white')  # or fill for solid shape
        ax.fill(hull_pts[:, 1], hull_pts[:, 0], facecolor="white", edgecolor="white", linewidth=1)

        # ax.plot(hull_pts[:, 1], hull_pts[:, 0], linewidth=1, color="white")

ax.set_title(f'Convex Hulls for {num_objects} Objects')
ax.axis('off')
plt.show()
```

```python
from rembg import remove
from PIL import Image
import numpy as np
from skimage import measure
from scipy.ndimage import label
import svgwrite
import io

# Step 1: Load image and remove background
input_path = 'test/images/test6.jpg'

img = Image.open(input_path).convert('RGBA')

img_bytes = io.BytesIO()
img.save(img_bytes, format='PNG')
img_bytes = img_bytes.getvalue()

output_bytes = remove(img_bytes)
output_img = Image.open(io.BytesIO(output_bytes)).convert('RGBA')

# Step 2: Threshold alpha channel to create binary mask
alpha = np.array(output_img.split()[-1])
binary_mask = alpha > 0

# Step 3: Label connected components
labeled, num_objects = label(binary_mask)

# Step 4: Create SVG canvas
height, width = binary_mask.shape
dwg = svgwrite.Drawing('output.svg', size=(f'{width}px', f'{height}px'))
dwg.viewbox(0, 0, width, height)

# Step 5: Extract contours and write to SVG
for obj_id in range(1, num_objects + 1):
    mask = labeled == obj_id
    contours = measure.find_contours(mask, level=0.5)

    for contour in contours:
        # Flip from (row, col) to (x, y)
        points = [(x, y) for y, x in contour]
        # Convert to SVG path string
        path_data = f'M {points[0][0]:.2f},{points[0][1]:.2f} ' + \
                    ' '.join(f'L {pt[0]:.2f},{pt[1]:.2f}' for pt in points[1:]) + ' Z'
        # Add to SVG
        dwg.add(dwg.path(d=path_data, fill='none', stroke='black', stroke_width=1))

# Step 6: Save SVG file
dwg.save()
print("✅ SVG saved as output.svg")
```

```python
from rembg import remove
from PIL import Image
import matplotlib.pyplot as plt
import numpy as np
from skimage import measure
from scipy.spatial import ConvexHull
from scipy.ndimage import label
from scipy.interpolate import splprep, splev
import io

# --- Step 1: Load and remove background ---
input_path = 'test/images/test6.jpg'
img = Image.open(input_path).convert('RGBA')

img_bytes = io.BytesIO()
img.save(img_bytes, format='PNG')
img_bytes = img_bytes.getvalue()

output_bytes = remove(img_bytes)
output_img = Image.open(io.BytesIO(output_bytes)).convert('RGBA')

# --- Step 2: Threshold the alpha channel ---
alpha = np.array(output_img.split()[-1])
binary_mask = alpha > 0

# --- Step 3: Label connected components ---
labeled, num_objects = label(binary_mask)

# --- Step 4: Helper function to smooth a 2D point set using B-spline ---
def smooth_contour(points, smoothing=5.0, num_points=200):
    x, y = points[:, 0], points[:, 1]
    
    # Wrap points if not already closed
    if not np.allclose(points[0], points[-1]):
        x = np.append(x, x[0])
        y = np.append(y, y[0])
    
    try:
        tck, _ = splprep([x, y], s=smoothing, per=True)
        u_fine = np.linspace(0, 1, num_points)
        x_smooth, y_smooth = splev(u_fine, tck)
        return np.vstack([x_smooth, y_smooth]).T
    except:
        # fallback: return original if splprep fails
        return points

# --- Step 5: Plot results ---
fig, ax = plt.subplots(figsize=(8, 8))
ax.imshow(binary_mask, cmap='gray')

for obj_id in range(1, num_objects + 1):
    mask = labeled == obj_id
    contours = measure.find_contours(mask, level=0.5)
    
    if not contours:
        continue

    # Merge all contours of this object
    points = np.vstack(contours)
    
    # Compute convex hull
    if len(points) >= 3:
        hull = ConvexHull(points)
        hull_pts = points[hull.vertices]
        hull_pts = np.vstack([hull_pts, hull_pts[0]])  # close the loop

        # Fill the convex hull with white
        ax.fill(hull_pts[:, 1], hull_pts[:, 0], facecolor="white", edgecolor="white", linewidth=1)


ax.set_title(f'Smoothed Convex Hulls ({num_objects} objects)')
ax.axis('off')
plt.tight_layout()
plt.show()
```

```python
from rembg import remove
from PIL import Image
import numpy as np
from skimage import measure
from scipy.ndimage import label
from scipy.spatial import ConvexHull
from rdp import rdp
import svgwrite
import io

# Load image and remove background
input_path = 'test/images/test6.jpg'
img = Image.open(input_path).convert('RGBA')
img_bytes = io.BytesIO()
img.save(img_bytes, format='PNG')
img_bytes = img_bytes.getvalue()
output_bytes = remove(img_bytes)
output_img = Image.open(io.BytesIO(output_bytes)).convert('RGBA')

# Convert to binary alpha mask
alpha = np.array(output_img.split()[-1])
binary_mask = alpha > 0

# Label connected components
labeled, num_objects = label(binary_mask)

# Setup SVG canvas
height, width = binary_mask.shape
dwg = svgwrite.Drawing('output_rdp.svg', size=(f'{width}px', f'{height}px'))
dwg.viewbox(0, 0, width, height)

# Loop through each object
for obj_id in range(1, num_objects + 1):
    mask = labeled == obj_id
    contours = measure.find_contours(mask, level=0.5)

    if not contours:
        continue

    # Merge all contours to get full shape
    points = np.vstack(contours)

    # Compute convex hull if enough points
    if len(points) >= 3:
        hull = ConvexHull(points)
        hull_pts = points[hull.vertices]

        # Close the loop for visual consistency
        hull_pts = np.vstack([hull_pts, hull_pts[0]])

        # Simplify using RDP
        simplified = rdp(hull_pts.tolist(), epsilon=2.0)
        simplified = np.array(simplified)

        # Convert to SVG path string
        path_data = f'M {simplified[0][1]:.2f},{simplified[0][0]:.2f} ' + \
                    ' '.join(f'L {pt[1]:.2f},{pt[0]:.2f}' for pt in simplified[1:]) + ' Z'

        # Add filled shape with white
        dwg.add(dwg.path(
            d=path_data,
            fill='white',
            stroke='black',
            stroke_width=2
        ))

# Save final SVG
dwg.save()
print("✅ Simplified convex hulls saved to output_rdp.svg")
```

```python
# Simple example: Invert PNG and convert to SVG

import numpy as np
from skimage import io, util
import vtracer
import tempfile
import os
```

```python

# Your input and output files
input_png = "test/images/output_example_1.png"  # Change this to your PNG file path
output_svg = "test/images/inverted_output.svg"  # Output SVG file

# Load the image
img = io.imread(input_png)
print(f"Loaded image shape: {img.shape}")

# Invert the image (black becomes white, white becomes black)
inverted_img = util.invert(img)

# Save inverted image to temporary file (vtracer needs a file path)
with tempfile.NamedTemporaryFile(suffix='.png', delete=False) as temp_file:
    temp_png_path = temp_file.name
    io.imsave(temp_png_path, inverted_img)

try:
    # Convert inverted image to SVG
    vtracer.convert_image_to_svg_py(temp_png_path, output_svg)
    print(f"Success! SVG saved as: {output_svg}")
    
finally:
    # Clean up temporary file
    if os.path.exists(temp_png_path):
        os.unlink(temp_png_path)

print("Done!")
```

```python

input_png = "test/images/output_example_2.png"  # Change this to your PNG file path
output_svg = "test/images/inverted_output2.svg" 


def convert_to_vector(input_image,output_svg):
    # open image 
    img = Image.open(input_png)

    # convert to numpy array for inversion
    img_array = np.array(img)

    # invert colors using numpy - subtract from max value (255 for 8-bit)
    if img_array.dtype == np.uint8:
        inverted_array = 255 - img_array
    else:
        inverted_array = np.iinfo(img_array.dtype).max - img_array

    # convert back to PIL Image
    inverted_img = Image.fromarray(inverted_array)

    # save in temp file
    with tempfile.NamedTemporaryFile(suffix='.png', delete=False) as temp_file:
        temp_png_path = temp_file.name
        inverted_img.save(temp_png_path)

    # send temp file to convert_image_to_svg_py
    vtracer.convert_image_to_svg_py(temp_png_path, output_svg, colormode='binary')

    # cleanup temp file
    os.unlink(temp_png_path)

convert_to_vector(input_png,output_svg)

# using scikit image add the Savitzky–Golay Filter 
# save the output as inverted_output_savitzky_golay.svg
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image, ImageEnhance
import skimage
from skimage import exposure, filters, morphology
from scipy import ndimage
```

```python

# Set up matplotlib for better display
plt.rcParams['figure.figsize'] = (15, 10)

def load_image(image_path):
    """Load image and convert to RGB if needed"""
    image = cv2.imread(image_path)
    if image is None:
        raise ValueError(f"Could not load image from {image_path}")
    return cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

def method1_histogram_equalization(image):
    """Method 1: Histogram Equalization to reduce shadow effects"""
    # Convert to LAB color space
    lab = cv2.cvtColor(image, cv2.COLOR_RGB2LAB)
    
    # Apply CLAHE (Contrast Limited Adaptive Histogram Equalization) to L channel
    clahe = cv2.createCLAHE(clipLimit=3.0, tileGridSize=(8,8))
    lab[:,:,0] = clahe.apply(lab[:,:,0])
    
    # Convert back to RGB
    enhanced = cv2.cvtColor(lab, cv2.COLOR_LAB2RGB)
    return enhanced

def method2_gamma_correction(image, gamma=0.5):
    """Method 2: Gamma correction to brighten shadows"""
    # Normalize to 0-1 range
    normalized = image.astype(np.float32) / 255.0
    
    # Apply gamma correction
    corrected = np.power(normalized, gamma)
    
    # Convert back to 0-255 range
    return (corrected * 255).astype(np.uint8)

def method3_shadow_removal_illumination(image):
    """Method 3: Shadow removal using illumination estimation"""
    # Convert to grayscale for processing
    gray = cv2.cvtColor(image, cv2.COLOR_RGB2GRAY)
    
    # Estimate illumination using morphological operations
    kernel = cv2.getStructuringElement(cv2.MORPH_ELLIPSE, (20, 20))
    illumination = cv2.morphologyEx(gray, cv2.MORPH_CLOSE, kernel)
    
    # Apply Gaussian blur to smooth illumination
    illumination = cv2.GaussianBlur(illumination, (51, 51), 0)
    
    # Normalize illumination
    illumination = illumination.astype(np.float32) / 255.0
    illumination = np.clip(illumination, 0.1, 1.0)  # Avoid division by zero
    
    # Apply correction to each channel
    corrected = np.zeros_like(image, dtype=np.float32)
    for i in range(3):
        channel = image[:,:,i].astype(np.float32) / 255.0
        corrected[:,:,i] = channel / illumination
    
    # Normalize and convert back
    corrected = np.clip(corrected, 0, 1)
    return (corrected * 255).astype(np.uint8)

def method4_retinex_enhancement(image):
    """Method 4: Single Scale Retinex for shadow reduction"""
    # Convert to float
    image_float = image.astype(np.float32) / 255.0
    
    # Apply Gaussian blur
    sigma = 80
    blurred = cv2.GaussianBlur(image_float, (0, 0), sigma)
    
    # Apply Retinex formula: log(I) - log(I_blurred)
    retinex = np.log(image_float + 1e-6) - np.log(blurred + 1e-6)
    
    # Normalize to 0-1 range
    retinex = (retinex - retinex.min()) / (retinex.max() - retinex.min())
    
    return (retinex * 255).astype(np.uint8)

def method5_adaptive_threshold_preprocessing(image):
    """Method 5: Preprocessing specifically for better bitmask creation"""
    # Convert to LAB
    lab = cv2.cvtColor(image, cv2.COLOR_RGB2LAB)
    
    # Enhance L channel
    l_channel = lab[:,:,0]
    
    # Apply bilateral filter to reduce noise while preserving edges
    filtered = cv2.bilateralFilter(l_channel, 9, 75, 75)
    
    # Apply adaptive histogram equalization
    clahe = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(8,8))
    enhanced_l = clahe.apply(filtered)
    
    # Put back enhanced L channel
    lab[:,:,0] = enhanced_l
    
    # Convert back to RGB
    result = cv2.cvtColor(lab, cv2.COLOR_LAB2RGB)
    
    return result

def create_bitmask(image, method='otsu', blur_kernel=5):
    """Create bitmask from processed image"""
    # Convert to grayscale
    gray = cv2.cvtColor(image, cv2.COLOR_RGB2GRAY)
    
    # Apply slight blur to reduce noise
    if blur_kernel > 0:
        gray = cv2.GaussianBlur(gray, (blur_kernel, blur_kernel), 0)
    
    if method == 'otsu':
        # Otsu's thresholding
        _, mask = cv2.threshold(gray, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)
    elif method == 'adaptive':
        # Adaptive thresholding
        mask = cv2.adaptiveThreshold(gray, 255, cv2.ADAPTIVE_THRESH_GAUSSIAN_C, 
                                   cv2.THRESH_BINARY, 11, 2)
    elif method == 'mean':
        # Simple mean thresholding
        thresh_val = np.mean(gray)
        _, mask = cv2.threshold(gray, thresh_val, 255, cv2.THRESH_BINARY)
    
    return mask

def clean_bitmask(mask, min_area=100):
    """Clean up bitmask by removing small objects and filling holes"""
    # Remove small objects
    cleaned = morphology.remove_small_objects(mask > 0, min_size=min_area)
    
    # Fill small holes
    cleaned = morphology.remove_small_holes(cleaned, area_threshold=min_area)
    
    return (cleaned * 255).astype(np.uint8)

def compare_methods(image_path):
    """Compare all shadow reduction methods"""
    # Load original image
    original = load_image(image_path)
    
    # Apply different methods
    methods = {
        'Original': original,
        'Histogram Eq': method1_histogram_equalization(original),
        'Gamma Correction': method2_gamma_correction(original, gamma=0.6),
        'Illumination': method3_shadow_removal_illumination(original),
        'Retinex': method4_retinex_enhancement(original),
        'Adaptive Prep': method5_adaptive_threshold_preprocessing(original)
    }
    
    # Create subplots
    fig, axes = plt.subplots(2, 3, figsize=(18, 12))
    axes = axes.ravel()
    
    for idx, (name, img) in enumerate(methods.items()):
        axes[idx].imshow(img)
        axes[idx].set_title(f'{name}', fontsize=12)
        axes[idx].axis('off')
    
    plt.tight_layout()
    plt.show()
    
    return methods

def compare_bitmasks(methods_dict, threshold_method='adaptive'):
    """Compare bitmasks created from different preprocessing methods"""
    fig, axes = plt.subplots(2, 3, figsize=(18, 12))
    axes = axes.ravel()
    
    for idx, (name, img) in enumerate(methods_dict.items()):
        # Create bitmask
        mask = create_bitmask(img, method=threshold_method)
        mask = clean_bitmask(mask)
        
        axes[idx].imshow(mask, cmap='gray')
        axes[idx].set_title(f'{name} - Bitmask', fontsize=12)
        axes[idx].axis('off')
    
    plt.tight_layout()
    plt.show()

# Example usage and demonstration
def demonstrate_shadow_reduction():
    """Demonstrate the complete workflow"""
    print("Shadow Reduction for Bitmask Creation Demo")
    print("=" * 50)
    
    # Note: Replace 'your_image_path.jpg' with actual image path
    image_path = 'test/images/test6.jpg'
    
    try:
        # Load and display original image
        original = load_image(image_path)
        
        plt.figure(figsize=(10, 6))
        plt.imshow(original)
        plt.title('Original Image')
        plt.axis('off')
        plt.show()
        
        # Compare all methods
        print("\n1. Comparing shadow reduction methods...")
        methods = compare_methods(image_path)
        
        print("\n2. Comparing resulting bitmasks...")
        compare_bitmasks(methods, threshold_method='adaptive')
        
        # Demonstrate the best method (usually adaptive preprocessing)
        print("\n3. Final result with best method:")
        best_processed = method5_adaptive_threshold_preprocessing(original)
        final_mask = create_bitmask(best_processed, method='adaptive')
        final_mask = clean_bitmask(final_mask)
        
        fig, (ax1, ax2, ax3) = plt.subplots(1, 3, figsize=(15, 5))
        
        ax1.imshow(original)
        ax1.set_title('Original')
        ax1.axis('off')
        
        ax2.imshow(best_processed)
        ax2.set_title('Shadow Reduced')
        ax2.axis('off')
        
        ax3.imshow(final_mask, cmap='gray')
        ax3.set_title('Final Bitmask')
        ax3.axis('off')
        
        plt.tight_layout()
        plt.show()
        
    except Exception as e:
        print(f"Error: {e}")
        print("Please ensure you have an image file and update the image_path variable")

# Additional utility functions
def save_processed_image(image, output_path):
    """Save processed image"""
    if image.dtype == np.float32:
        image = (image * 255).astype(np.uint8)
    
    # Convert RGB to BGR for OpenCV saving
    bgr_image = cv2.cvtColor(image, cv2.COLOR_RGB2BGR)
    cv2.imwrite(output_path, bgr_image)
    print(f"Saved processed image to: {output_path}")

def batch_process_images(input_folder, output_folder, method_func=method5_adaptive_threshold_preprocessing):
    """Process multiple images in batch"""
    import os
    from pathlib import Path
    
    input_path = Path(input_folder)
    output_path = Path(output_folder)
    output_path.mkdir(exist_ok=True)
    
    image_extensions = ['.jpg', '.jpeg', '.png', '.bmp', '.tiff']
    
    for img_file in input_path.iterdir():
        if img_file.suffix.lower() in image_extensions:
            try:
                # Load and process image
                image = load_image(str(img_file))
                processed = method_func(image)
                
                # Save processed image
                output_file = output_path / f"processed_{img_file.name}"
                save_processed_image(processed, str(output_file))
                
                # Create and save bitmask
                mask = create_bitmask(processed, method='adaptive')
                mask = clean_bitmask(mask)
                mask_file = output_path / f"mask_{img_file.stem}.png"
                cv2.imwrite(str(mask_file), mask)
                
                print(f"Processed: {img_file.name}")
                
            except Exception as e:
                print(f"Error processing {img_file.name}: {e}")

# Run demonstration (uncomment to use)
demonstrate_shadow_reduction()

print("Shadow reduction notebook loaded successfully!")
print("\nAvailable functions:")
print("- method1_histogram_equalization(image)")
print("- method2_gamma_correction(image, gamma=0.5)")
print("- method3_shadow_removal_illumination(image)")
print("- method4_retinex_enhancement(image)")
print("- method5_adaptive_threshold_preprocessing(image)")
print("- create_bitmask(image, method='otsu')")
print("- clean_bitmask(mask)")
print("- compare_methods(image_path)")
print("- demonstrate_shadow_reduction()")
print("\nTo use: Update 'image_path' variable and run demonstrate_shadow_reduction()")
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image, ImageEnhance

def otsu_threshold(image):
    """OpenCV implementation of Otsu's threshold"""
    if len(image.shape) == 3:
        gray = cv2.cvtColor(image, cv2.COLOR_RGB2GRAY)
    else:
        gray = image
    
    threshold_value, _ = cv2.threshold(gray, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)
    return threshold_value

def load_image(image_path):
    """Load image using OpenCV"""
    img = cv2.imread(image_path)
    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    return img_rgb

def brighten_shadows_gamma(image, gamma=0.5):
    """Brighten shadows using gamma correction"""
    # Normalize image to 0-1 range
    normalized = image.astype(np.float32) / 255.0
    
    # Apply gamma correction (gamma < 1 brightens shadows)
    corrected = np.power(normalized, gamma)
    
    # Convert back to 0-255 range
    result = (corrected * 255).astype(np.uint8)
    return result

def brighten_shadows_clahe(image, clip_limit=2.0, tile_grid_size=(8,8)):
    """Brighten shadows using CLAHE (Contrast Limited Adaptive Histogram Equalization)"""
    # Convert to LAB color space
    lab = cv2.cvtColor(image, cv2.COLOR_RGB2LAB)
    
    # Apply CLAHE to L channel (lightness)
    clahe = cv2.createCLAHE(clipLimit=clip_limit, tileGridSize=tile_grid_size)
    lab[:,:,0] = clahe.apply(lab[:,:,0])
    
    # Convert back to RGB
    result = cv2.cvtColor(lab, cv2.COLOR_LAB2RGB)
    return result

def brighten_shadows_exposure(image, shadow_strength=0.5):
    """Brighten shadows using exposure adjustment"""
    # Convert to float
    img_float = image.astype(np.float32) / 255.0
    
    # Create shadow mask (darker areas)
    gray = cv2.cvtColor(image, cv2.COLOR_RGB2GRAY)
    shadow_mask = (gray < 128).astype(np.float32)
    
    # Apply brightness boost to shadows
    brightness_boost = shadow_strength * shadow_mask[:,:,np.newaxis]
    result = img_float + brightness_boost
    
    # Clip values and convert back
    result = np.clip(result, 0, 1)
    return (result * 255).astype(np.uint8)

def create_bitmask_from_brightness(image, threshold_value=None, invert=False):
    """Create bitmask based on brightness levels"""
    # Convert to grayscale
    gray = cv2.cvtColor(image, cv2.COLOR_RGB2GRAY)
    
    # Auto-determine threshold if not provided
    if threshold_value is None:
        threshold_value = otsu_threshold(image)
    
    # Create binary mask
    if invert:
        mask = (gray < threshold_value).astype(np.uint8) * 255
    else:
        mask = (gray > threshold_value).astype(np.uint8) * 255
    
    return mask, threshold_value

def refine_bitmask(mask, kernel_size=3, iterations=1):
    """Refine bitmask using morphological operations"""
    kernel = np.ones((kernel_size, kernel_size), np.uint8)
    
    # Remove noise
    cleaned = cv2.morphologyEx(mask, cv2.MORPH_CLOSE, kernel, iterations=iterations)
    cleaned = cv2.morphologyEx(cleaned, cv2.MORPH_OPEN, kernel, iterations=iterations)
    
    return cleaned

def process_image_pipeline(image_path, method='clahe', show_plots=True):
    """Complete pipeline for shadow brightening and bitmask creation"""
    
    # Load image
    original = load_image(image_path)
    
    # Brighten shadows using selected method
    if method == 'gamma':
        brightened = brighten_shadows_gamma(original, gamma=0.4)
    elif method == 'clahe':
        brightened = brighten_shadows_clahe(original, clip_limit=3.0)
    elif method == 'exposure':
        brightened = brighten_shadows_exposure(original, shadow_strength=0.3)
    else:
        raise ValueError("Method must be 'gamma', 'clahe', or 'exposure'")
    
    # Create bitmask
    mask, threshold = create_bitmask_from_brightness(brightened)
    
    # Refine mask
    refined_mask = refine_bitmask(mask)
    
    if show_plots:
        # Display results
        fig, axes = plt.subplots(2, 3, figsize=(15, 10))
        
        axes[0,0].imshow(original)
        axes[0,0].set_title('Original Image')
        axes[0,0].axis('off')
        
        axes[0,1].imshow(brightened)
        axes[0,1].set_title(f'Shadow Brightened ({method})')
        axes[0,1].axis('off')
        
        # Show histograms
        axes[0,2].hist(cv2.cvtColor(original, cv2.COLOR_RGB2GRAY).flatten(), 
                      bins=50, alpha=0.7, label='Original', color='blue')
        axes[0,2].hist(cv2.cvtColor(brightened, cv2.COLOR_RGB2GRAY).flatten(), 
                      bins=50, alpha=0.7, label='Brightened', color='red')
        axes[0,2].set_title('Brightness Histograms')
        axes[0,2].legend()
        
        axes[1,0].imshow(mask, cmap='gray')
        axes[1,0].set_title(f'Initial Bitmask (threshold: {threshold:.1f})')
        axes[1,0].axis('off')
        
        axes[1,1].imshow(refined_mask, cmap='gray')
        axes[1,1].set_title('Refined Bitmask')
        axes[1,1].axis('off')
        
        # Show mask overlay
        overlay = original.copy()
        overlay[refined_mask == 255] = [255, 0, 0]  # Red overlay for mask areas
        axes[1,2].imshow(cv2.addWeighted(original, 0.7, overlay, 0.3, 0))
        axes[1,2].set_title('Mask Overlay on Original')
        axes[1,2].axis('off')
        
        plt.tight_layout()
        plt.show()
    
    return original, brightened, refined_mask

# Example usage:
"""
# Load and process an image
image_path = 'your_image.jpg'

# Try different methods
original, brightened, mask = process_image_pipeline(image_path, method='clahe')

# Save results
cv2.imwrite('brightened_image.jpg', cv2.cvtColor(brightened, cv2.COLOR_RGB2BGR))
cv2.imwrite('bitmask.png', mask)

# Alternative methods to try:
# process_image_pipeline(image_path, method='gamma')
# process_image_pipeline(image_path, method='exposure')
"""

# Advanced bitmask creation with multiple thresholds
def create_multi_level_mask(image, num_levels=3):
    """Create multi-level bitmask with different brightness zones"""
    gray = cv2.cvtColor(image, cv2.COLOR_RGB2GRAY)
    
    # Create multiple threshold levels
    masks = []
    for i in range(num_levels):
        threshold = (i + 1) * (255 // (num_levels + 1))
        mask = (gray > threshold).astype(np.uint8) * 255
        masks.append(mask)
    
    return masks

# Interactive threshold adjustment
def interactive_threshold_adjustment(image):
    """Create interactive widget for threshold adjustment (requires ipywidgets)"""
    from ipywidgets import interact, IntSlider
    
    def update_mask(threshold_val):
        mask, _ = create_bitmask_from_brightness(image, threshold_val)
        
        fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))
        ax1.imshow(image)
        ax1.set_title('Original Image')
        ax1.axis('off')
        
        ax2.imshow(mask, cmap='gray')
        ax2.set_title(f'Bitmask (threshold: {threshold_val})')
        ax2.axis('off')
        
        plt.show()
    
    interact(update_mask, threshold_val=IntSlider(min=0, max=255, step=5, value=128))
```

```python
# Image Shadow Brightening Techniques
# Install required packages: pip install opencv-python pillow matplotlib numpy scikit-image

import cv2
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image, ImageEnhance
from skimage import exposure, filters
import os

def load_image(image_path):
    """Load image using OpenCV and convert to RGB"""
    img = cv2.imread(image_path)
    if img is None:
        raise ValueError(f"Could not load image from {image_path}")
    return cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

def method1_gamma_correction(image, gamma=0.6):
    """Brighten shadows using gamma correction (lower gamma brightens shadows)"""
    # Normalize to 0-1 range
    normalized = image.astype(np.float32) / 255.0
    # Apply gamma correction
    corrected = np.power(normalized, gamma)
    # Convert back to 0-255 range
    return (corrected * 255).astype(np.uint8)

def method2_histogram_equalization(image):
    """Brighten shadows using adaptive histogram equalization"""
    # Convert to LAB color space
    lab = cv2.cvtColor(image, cv2.COLOR_RGB2LAB)
    l, a, b = cv2.split(lab)
    
    # Apply CLAHE (Contrast Limited Adaptive Histogram Equalization) to L channel
    clahe = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(8,8))
    l_clahe = clahe.apply(l)
    
    # Merge channels and convert back to RGB
    lab_clahe = cv2.merge([l_clahe, a, b])
    return cv2.cvtColor(lab_clahe, cv2.COLOR_LAB2RGB)

def method3_shadow_highlight_adjustment(image, shadow_amount=0.7, highlight_amount=0.0):
    """Selective shadow/highlight adjustment"""
    # Convert to float32
    img_float = image.astype(np.float32) / 255.0
    
    # Create shadow and highlight masks
    luminance = cv2.cvtColor(image, cv2.COLOR_RGB2GRAY).astype(np.float32) / 255.0
    shadow_mask = 1.0 - luminance  # Darker areas get higher values
    highlight_mask = luminance     # Brighter areas get higher values
    
    # Apply Gaussian blur to masks for smooth transitions
    shadow_mask = cv2.GaussianBlur(shadow_mask, (0, 0), 30)
    highlight_mask = cv2.GaussianBlur(highlight_mask, (0, 0), 30)
    
    # Adjust shadows and highlights
    shadow_adjustment = shadow_mask * shadow_amount
    highlight_adjustment = highlight_mask * highlight_amount
    
    # Apply adjustments to each channel
    result = img_float.copy()
    for i in range(3):  # RGB channels
        result[:,:,i] = np.clip(result[:,:,i] + shadow_adjustment - highlight_adjustment, 0, 1)
    
    return (result * 255).astype(np.uint8)

def method4_unsharp_masking_shadows(image):
    """Brighten shadows using unsharp masking technique"""
    # Convert to LAB
    lab = cv2.cvtColor(image, cv2.COLOR_RGB2LAB).astype(np.float32)
    l, a, b = cv2.split(lab)
    
    # Create blurred version of L channel
    l_blur = cv2.GaussianBlur(l, (0, 0), 10)
    
    # Create mask for shadows (invert so shadows are bright)
    shadow_mask = 255 - l
    shadow_mask = shadow_mask / 255.0
    
    # Apply shadow brightening
    l_enhanced = l + (shadow_mask * 30)  # Adjust the 30 for strength
    l_enhanced = np.clip(l_enhanced, 0, 255)
    
    # Merge and convert back
    lab_enhanced = cv2.merge([l_enhanced, a, b]).astype(np.uint8)
    return cv2.cvtColor(lab_enhanced, cv2.COLOR_LAB2RGB)

def method5_log_transform(image):
    """Brighten shadows using logarithmic transformation"""
    # Convert to float and normalize
    img_float = image.astype(np.float32)
    
    # Apply log transform (c * log(1 + image))
    c = 255 / np.log(1 + np.max(img_float))
    log_transformed = c * np.log(1 + img_float)
    
    return np.clip(log_transformed, 0, 255).astype(np.uint8)

def compare_methods(image_path):
    """Compare all shadow brightening methods"""
    # Load original image
    original = load_image(image_path)
    
    # Apply different methods
    gamma_corrected = method1_gamma_correction(original)
    hist_equalized = method2_histogram_equalization(original)
    shadow_adjusted = method3_shadow_highlight_adjustment(original)
    unsharp_masked = method4_unsharp_masking_shadows(original)
    log_transformed = method5_log_transform(original)
    
    # Create comparison plot
    fig, axes = plt.subplots(2, 3, figsize=(18, 12))
    
    images = [original, gamma_corrected, hist_equalized, 
              shadow_adjusted, unsharp_masked, log_transformed]
    titles = ['Original', 'Gamma Correction', 'Histogram Equalization',
              'Shadow/Highlight Adj.', 'Unsharp Masking', 'Log Transform']
    
    for i, (img, title) in enumerate(zip(images, titles)):
        row, col = i // 3, i % 3
        axes[row, col].imshow(img)
        axes[row, col].set_title(title, fontsize=12)
        axes[row, col].axis('off')
    
    plt.tight_layout()
    plt.show()
    
    return {
        'original': original,
        'gamma_corrected': gamma_corrected,
        'hist_equalized': hist_equalized,
        'shadow_adjusted': shadow_adjusted,
        'unsharp_masked': unsharp_masked,
        'log_transformed': log_transformed
    }

# Example usage:
# Replace 'your_image.jpg' with the path to your image
IMAGE_PATH = 'test/images/test6.jpg'

try:
    # Compare all methods
    results = compare_methods(IMAGE_PATH)
    
    # You can also apply individual methods:
    original = load_image(IMAGE_PATH)
    
    # Method 1: Quick gamma correction (most common)
    brightened = method1_gamma_correction(original, gamma=0.5)
    
    # Method 3: More advanced shadow/highlight adjustment
    advanced_brightened = method3_shadow_highlight_adjustment(
        original, shadow_amount=0.8, highlight_amount=0.1
    )
    
    # Display individual result
    plt.figure(figsize=(15, 5))
    plt.subplot(1, 3, 1)
    plt.imshow(original)
    plt.title('Original')
    plt.axis('off')
    
    plt.subplot(1, 3, 2)
    plt.imshow(brightened)
    plt.title('Gamma Correction')
    plt.axis('off')
    
    plt.subplot(1, 3, 3)
    plt.imshow(advanced_brightened)
    plt.title('Shadow/Highlight Adjustment')
    plt.axis('off')
    
    plt.tight_layout()
    plt.show()
    
    # Save the results
    cv2.imwrite('brightened_gamma.jpg', cv2.cvtColor(brightened, cv2.COLOR_RGB2BGR))
    cv2.imwrite('brightened_advanced.jpg', cv2.cvtColor(advanced_brightened, cv2.COLOR_RGB2BGR))
    
    print("Images saved successfully!")
    
except Exception as e:
    print(f"Error: {e}")
    print("Please make sure to:")
    print("1. Install required packages: pip install opencv-python pillow matplotlib numpy scikit-image")
    print("2. Replace 'your_image.jpg' with the actual path to your image")

# Interactive parameter adjustment function
def interactive_shadow_adjustment(image_path):
    """Interactive function to adjust shadow brightening parameters"""
    original = load_image(image_path)
    
    def adjust_shadows(gamma=0.6, shadow_amount=0.7, highlight_amount=0.0):
        gamma_result = method1_gamma_correction(original, gamma)
        shadow_result = method3_shadow_highlight_adjustment(original, shadow_amount, highlight_amount)
        
        plt.figure(figsize=(15, 5))
        plt.subplot(1, 3, 1)
        plt.imshow(original)
        plt.title('Original')
        plt.axis('off')
        
        plt.subplot(1, 3, 2)
        plt.imshow(gamma_result)
        plt.title(f'Gamma: {gamma}')
        plt.axis('off')
        
        plt.subplot(1, 3, 3)
        plt.imshow(shadow_result)
        plt.title(f'Shadow: {shadow_amount}, Highlight: {highlight_amount}')
        plt.axis('off')
        
        plt.tight_layout()
        plt.show()
        
        return gamma_result, shadow_result
    
    # You can call this function with different parameters
    # adjust_shadows(gamma=0.5, shadow_amount=0.8, highlight_amount=0.1)
    
    return adjust_shadows
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.widgets import Slider
import cv2
from PIL import Image
import io

# Load and prepare the image
def load_image(image_path):
    """Load image using PIL and convert to numpy array"""
    img = Image.open(image_path)
    img_array = np.array(img)
    if len(img_array.shape) == 3 and img_array.shape[2] == 3:
        # Convert RGB to BGR for OpenCV operations
        img_array = cv2.cvtColor(img_array, cv2.COLOR_RGB2BGR)
    return img_array

def adjust_shadows(image, shadow_lift=0, highlight_compress=0, gamma=1.0):
    """
    Adjust shadows and highlights in an image
    
    Parameters:
    - shadow_lift: Amount to brighten shadows (0-100)
    - highlight_compress: Amount to compress highlights (0-100) 
    - gamma: Gamma correction (0.1-3.0)
    """
    # Convert to float32 for processing
    img_float = image.astype(np.float32) / 255.0
    
    # Convert to LAB color space for better shadow/highlight control
    if len(img_float.shape) == 3:
        lab = cv2.cvtColor(img_float, cv2.COLOR_BGR2LAB)
        l_channel = lab[:, :, 0] / 100.0  # Normalize L channel
        a_channel = lab[:, :, 1]
        b_channel = lab[:, :, 2]
    else:
        l_channel = img_float
    
    # Shadow lifting: brighten dark areas more than bright areas
    shadow_mask = 1 - l_channel  # Inverted luminance (shadows = high values)
    shadow_adjustment = (shadow_lift / 100.0) * shadow_mask * 0.3
    
    # Highlight compression: darken bright areas
    highlight_mask = l_channel  # Luminance (highlights = high values)
    highlight_adjustment = -(highlight_compress / 100.0) * highlight_mask * 0.2
    
    # Apply adjustments
    l_adjusted = l_channel + shadow_adjustment + highlight_adjustment
    
    # Apply gamma correction
    l_adjusted = np.power(np.clip(l_adjusted, 0, 1), 1/gamma)
    
    # Reconstruct image
    if len(img_float.shape) == 3:
        lab[:, :, 0] = l_adjusted * 100.0
        result = cv2.cvtColor(lab, cv2.COLOR_LAB2BGR)
    else:
        result = l_adjusted
    
    # Convert back to uint8
    result = np.clip(result * 255, 0, 255).astype(np.uint8)
    
    return result

# Interactive shadow adjustment tool
class ShadowAdjustmentTool:
    def __init__(self, image_path):
        self.original_image = load_image(image_path)
        self.current_image = self.original_image.copy()
        
        # Create the figure and subplots
        self.fig, (self.ax_original, self.ax_adjusted) = plt.subplots(1, 2, figsize=(15, 6))
        self.fig.suptitle('Interactive Shadow Brightening Tool', fontsize=16)
        
        # Display original image
        if len(self.original_image.shape) == 3:
            display_original = cv2.cvtColor(self.original_image, cv2.COLOR_BGR2RGB)
        else:
            display_original = self.original_image
        self.ax_original.imshow(display_original, cmap='gray' if len(self.original_image.shape) == 2 else None)
        self.ax_original.set_title('Original Image')
        self.ax_original.axis('off')
        
        # Display adjusted image (initially same as original)
        self.im_adjusted = self.ax_adjusted.imshow(display_original, cmap='gray' if len(self.original_image.shape) == 2 else None)
        self.ax_adjusted.set_title('Adjusted Image')
        self.ax_adjusted.axis('off')
        
        # Create sliders
        plt.subplots_adjust(bottom=0.35)
        
        # Shadow lift slider
        ax_shadow = plt.axes([0.2, 0.25, 0.6, 0.03])
        self.shadow_slider = Slider(ax_shadow, 'Shadow Lift', 0, 100, valinit=0, valfmt='%d')
        
        # Highlight compress slider
        ax_highlight = plt.axes([0.2, 0.2, 0.6, 0.03])
        self.highlight_slider = Slider(ax_highlight, 'Highlight Compress', 0, 100, valinit=0, valfmt='%d')
        
        # Gamma slider
        ax_gamma = plt.axes([0.2, 0.15, 0.6, 0.03])
        self.gamma_slider = Slider(ax_gamma, 'Gamma', 0.1, 3.0, valinit=1.0, valfmt='%.2f')
        
        # Overall brightness slider
        ax_brightness = plt.axes([0.2, 0.1, 0.6, 0.03])
        self.brightness_slider = Slider(ax_brightness, 'Brightness', -50, 50, valinit=0, valfmt='%d')
        
        # Connect sliders to update function
        self.shadow_slider.on_changed(self.update_image)
        self.highlight_slider.on_changed(self.update_image)
        self.gamma_slider.on_changed(self.update_image)
        self.brightness_slider.on_changed(self.update_image)
        
        # Add reset button
        ax_reset = plt.axes([0.85, 0.05, 0.1, 0.04])
        from matplotlib.widgets import Button
        self.reset_button = Button(ax_reset, 'Reset')
        self.reset_button.on_clicked(self.reset_sliders)
        
    def update_image(self, val):
        """Update the displayed image based on slider values"""
        shadow_lift = self.shadow_slider.val
        highlight_compress = self.highlight_slider.val
        gamma = self.gamma_slider.val
        brightness = self.brightness_slider.val
        
        # Apply shadow/highlight adjustments
        adjusted = adjust_shadows(self.original_image, shadow_lift, highlight_compress, gamma)
        
        # Apply brightness adjustment
        if brightness != 0:
            adjusted = cv2.convertScaleAbs(adjusted, alpha=1.0, beta=brightness)
        
        # Update display
        if len(adjusted.shape) == 3:
            display_adjusted = cv2.cvtColor(adjusted, cv2.COLOR_BGR2RGB)
        else:
            display_adjusted = adjusted
            
        self.im_adjusted.set_array(display_adjusted)
        self.current_image = adjusted
        self.fig.canvas.draw()
        
    def reset_sliders(self, event):
        """Reset all sliders to default values"""
        self.shadow_slider.reset()
        self.highlight_slider.reset()
        self.gamma_slider.reset()
        self.brightness_slider.reset()
        
    def save_adjusted_image(self, output_path):
        """Save the current adjusted image"""
        if len(self.current_image.shape) == 3:
            # Convert BGR back to RGB for saving
            save_image = cv2.cvtColor(self.current_image, cv2.COLOR_BGR2RGB)
        else:
            save_image = self.current_image
        
        Image.fromarray(save_image).save(output_path)
        print(f"Adjusted image saved to: {output_path}")

# Usage example:
"""
# Load your image and create the interactive tool
image_path = "your_image.jpg"  # Replace with your image path
tool = ShadowAdjustmentTool(image_path)

# Show the interactive plot
plt.show()

# After adjusting, save the result
# tool.save_adjusted_image("adjusted_image.jpg")
"""

# Alternative: Simple function-based approach
def create_shadow_adjustment_tool(image_path):
    """
    Create an interactive shadow adjustment tool
    
    Usage:
    create_shadow_adjustment_tool("path/to/your/image.jpg")
    """
    
    # Load image
    original = load_image(image_path)
    
    # Create interactive plot
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 6))
    
    # Display original
    if len(original.shape) == 3:
        display_orig = cv2.cvtColor(original, cv2.COLOR_BGR2RGB)
    else:
        display_orig = original
    ax1.imshow(display_orig, cmap='gray' if len(original.shape) == 2 else None)
    ax1.set_title('Original')
    ax1.axis('off')
    
    # Initial adjusted display
    im2 = ax2.imshow(display_orig, cmap='gray' if len(original.shape) == 2 else None)
    ax2.set_title('Adjusted')
    ax2.axis('off')
    
    # Make space for sliders
    plt.subplots_adjust(bottom=0.3)
    
    # Create sliders
    ax_shadow = plt.axes([0.2, 0.2, 0.6, 0.03])
    shadow_slider = Slider(ax_shadow, 'Shadow Lift', 0, 100, valinit=20)
    
    ax_gamma = plt.axes([0.2, 0.15, 0.6, 0.03])
    gamma_slider = Slider(ax_gamma, 'Gamma', 0.1, 3.0, valinit=1.2)
    
    ax_bright = plt.axes([0.2, 0.1, 0.6, 0.03])
    bright_slider = Slider(ax_bright, 'Brightness', -50, 50, valinit=10)
    
    def update(val):
        adjusted = adjust_shadows(original, 
                                shadow_slider.val, 
                                0, 
                                gamma_slider.val)
        
        if bright_slider.val != 0:
            adjusted = cv2.convertScaleAbs(adjusted, alpha=1.0, beta=bright_slider.val)
        
        if len(adjusted.shape) == 3:
            display_adj = cv2.cvtColor(adjusted, cv2.COLOR_BGR2RGB)
        else:
            display_adj = adjusted
            
        im2.set_array(display_adj)
        fig.canvas.draw()
    
    shadow_slider.on_changed(update)
    gamma_slider.on_changed(update)
    bright_slider.on_changed(update)
    
    # Initial update to show some shadow lifting
    update(None)
    
    plt.show()
    
    return fig, (shadow_slider, gamma_slider, bright_slider)

# Example usage:
# fig, sliders = create_shadow_adjustment_tool("your_image.jpg")
```

```python
import numpy as np
import matplotlib.pyplot as plt
import cv2
from PIL import Image
import ipywidgets as widgets
from IPython.display import display
import io

# Set matplotlib to inline for compatibility
%matplotlib inline

# Load and prepare the image
def load_image(image_path):
    """Load image using PIL and convert to numpy array"""
    try:
        img = Image.open(image_path)
        img_array = np.array(img)
        if len(img_array.shape) == 3 and img_array.shape[2] == 3:
            # Convert RGB to BGR for OpenCV operations
            img_array = cv2.cvtColor(img_array, cv2.COLOR_RGB2BGR)
        return img_array
    except Exception as e:
        print(f"Error loading image: {e}")
        return None

def adjust_shadows(image, shadow_lift=0, highlight_compress=0, gamma=1.0, brightness=0):
    """
    Adjust shadows and highlights in an image
    
    Parameters:
    - shadow_lift: Amount to brighten shadows (0-100)
    - highlight_compress: Amount to compress highlights (0-100) 
    - gamma: Gamma correction (0.1-3.0)
    - brightness: Overall brightness adjustment (-50 to 50)
    """
    # Convert to float32 for processing
    img_float = image.astype(np.float32) / 255.0
    
    # Convert to LAB color space for better shadow/highlight control
    if len(img_float.shape) == 3:
        lab = cv2.cvtColor(img_float, cv2.COLOR_BGR2LAB)
        l_channel = lab[:, :, 0] / 100.0  # Normalize L channel
        a_channel = lab[:, :, 1]
        b_channel = lab[:, :, 2]
    else:
        l_channel = img_float
    
    # Shadow lifting: brighten dark areas more than bright areas
    shadow_mask = 1 - l_channel  # Inverted luminance (shadows = high values)
    shadow_adjustment = (shadow_lift / 100.0) * shadow_mask * 0.3
    
    # Highlight compression: darken bright areas
    highlight_mask = l_channel  # Luminance (highlights = high values)
    highlight_adjustment = -(highlight_compress / 100.0) * highlight_mask * 0.2
    
    # Apply adjustments
    l_adjusted = l_channel + shadow_adjustment + highlight_adjustment
    
    # Apply gamma correction
    l_adjusted = np.power(np.clip(l_adjusted, 0, 1), 1/gamma)
    
    # Reconstruct image
    if len(img_float.shape) == 3:
        lab[:, :, 0] = l_adjusted * 100.0
        result = cv2.cvtColor(lab, cv2.COLOR_LAB2BGR)
    else:
        result = l_adjusted
    
    # Apply brightness adjustment
    if brightness != 0:
        result = result + (brightness / 100.0)
    
    # Convert back to uint8
    result = np.clip(result * 255, 0, 255).astype(np.uint8)
    
    return result

# Interactive shadow adjustment tool using ipywidgets
class InteractiveShadowTool:
    def __init__(self, image_path):
        self.original_image = load_image(image_path)
        
        if self.original_image is None:
            print("Failed to load image. Please check the path.")
            return
            
        self.current_image = self.original_image.copy()
        
        # Create widgets
        self.shadow_slider = widgets.IntSlider(
            value=0, min=0, max=100, step=1,
            description='Shadow Lift:', style={'description_width': 'initial'}
        )
        
        self.highlight_slider = widgets.IntSlider(
            value=0, min=0, max=100, step=1,
            description='Highlight Compress:', style={'description_width': 'initial'}
        )
        
        self.gamma_slider = widgets.FloatSlider(
            value=1.0, min=0.1, max=3.0, step=0.1,
            description='Gamma:', style={'description_width': 'initial'}
        )
        
        self.brightness_slider = widgets.IntSlider(
            value=0, min=-50, max=50, step=1,
            description='Brightness:', style={'description_width': 'initial'}
        )
        
        self.reset_button = widgets.Button(
            description='Reset All',
            button_style='warning'
        )
        
        self.save_button = widgets.Button(
            description='Save Image',
            button_style='success'
        )
        
        # Create output widget for the plot
        self.output = widgets.Output()
        
        # Connect events
        self.shadow_slider.observe(self.update_image, names='value')
        self.highlight_slider.observe(self.update_image, names='value')
        self.gamma_slider.observe(self.update_image, names='value')
        self.brightness_slider.observe(self.update_image, names='value')
        self.reset_button.on_click(self.reset_values)
        self.save_button.on_click(self.save_image)
        
        # Initial display
        self.update_image()
        
        # Layout
        controls = widgets.VBox([
            self.shadow_slider,
            self.highlight_slider, 
            self.gamma_slider,
            self.brightness_slider,
            widgets.HBox([self.reset_button, self.save_button])
        ])
        
        self.widget = widgets.HBox([controls, self.output])
        
    def update_image(self, change=None):
        """Update the displayed image based on slider values"""
        with self.output:
            self.output.clear_output(wait=True)
            
            # Get current slider values
            shadow_lift = self.shadow_slider.value
            highlight_compress = self.highlight_slider.value
            gamma = self.gamma_slider.value
            brightness = self.brightness_slider.value
            
            # Apply adjustments
            adjusted = adjust_shadows(
                self.original_image, 
                shadow_lift, 
                highlight_compress, 
                gamma, 
                brightness
            )
            
            self.current_image = adjusted
            
            # Create side-by-side comparison
            fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 6))
            
            # Display original
            if len(self.original_image.shape) == 3:
                display_orig = cv2.cvtColor(self.original_image, cv2.COLOR_BGR2RGB)
            else:
                display_orig = self.original_image
            ax1.imshow(display_orig, cmap='gray' if len(self.original_image.shape) == 2 else None)
            ax1.set_title('Original Image')
            ax1.axis('off')
            
            # Display adjusted
            if len(adjusted.shape) == 3:
                display_adj = cv2.cvtColor(adjusted, cv2.COLOR_BGR2RGB)
            else:
                display_adj = adjusted
            ax2.imshow(display_adj, cmap='gray' if len(adjusted.shape) == 2 else None)
            ax2.set_title(f'Adjusted (Shadow:{shadow_lift}, Gamma:{gamma:.1f}, Brightness:{brightness})')
            ax2.axis('off')
            
            plt.tight_layout()
            plt.show()
            
    def reset_values(self, button):
        """Reset all sliders to default values"""
        self.shadow_slider.value = 0
        self.highlight_slider.value = 0
        self.gamma_slider.value = 1.0
        self.brightness_slider.value = 0
        
    def save_image(self, button):
        """Save the current adjusted image"""
        try:
            if len(self.current_image.shape) == 3:
                # Convert BGR back to RGB for saving
                save_image = cv2.cvtColor(self.current_image, cv2.COLOR_BGR2RGB)
            else:
                save_image = self.current_image
            
            output_path = "adjusted_shadow_image.jpg"
            Image.fromarray(save_image).save(output_path)
            print(f"✅ Image saved as: {output_path}")
        except Exception as e:
            print(f"❌ Error saving image: {e}")
    
    def display(self):
        """Display the interactive widget"""
        display(self.widget)

# Simple function for quick usage
def create_interactive_shadow_tool(image_path):
    """
    Create and display an interactive shadow adjustment tool
    
    Usage:
    create_interactive_shadow_tool("test/images/test6.jpg")
    """
    tool = InteractiveShadowTool(image_path)
    tool.display()
    return tool

# Usage with your image:
print("🎯 Creating interactive shadow brightening tool...")
print("📁 Loading image: test/images/test10.jpg")

# Create the tool
# tool = create_interactive_shadow_tool("test/images/test6.jpg")

print("\n🎛️ Use the sliders above to adjust:")
print("   • Shadow Lift: Brightens dark areas")
print("   • Highlight Compress: Prevents overexposure")  
print("   • Gamma: Adjusts contrast curve")
print("   • Brightness: Overall brightness")
print("   • Reset All: Returns to original")
print("   • Save Image: Saves as 'adjusted_shadow_image.jpg'")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image, ImageEnhance
import ipywidgets as widgets
from IPython.display import display
import cv2

# Load the image
image_path = 'test/images/test6.jpg'
original_image = Image.open(image_path)
original_array = np.array(original_image)

# Global variable to store current adjusted image
current_adjusted_image = None

def adjust_exposure_advanced(image_array, exposure, gamma=1.0):
    """
    Advanced exposure adjustment with gamma correction
    exposure: multiplication factor for brightness
    gamma: gamma correction value (1.0 = no correction)
    """
    # Convert to float for processing
    img_float = image_array.astype(np.float32) / 255.0
    
    # Apply exposure adjustment
    img_exposed = img_float * exposure
    
    # Apply gamma correction
    img_gamma = np.power(img_exposed, 1.0/gamma)
    
    # Clip values to valid range and convert back to uint8
    img_final = np.clip(img_gamma * 255, 0, 255).astype(np.uint8)
    
    return img_final

def update_advanced_image(exposure, gamma, contrast, saturation):
    """Update image with multiple adjustments"""
    global current_adjusted_image
    
    plt.clf()
    
    # Apply exposure and gamma
    adjusted_array = adjust_exposure_advanced(original_array, exposure, gamma)
    adjusted_image = Image.fromarray(adjusted_array)
    
    # Apply contrast adjustment
    contrast_enhancer = ImageEnhance.Contrast(adjusted_image)
    adjusted_image = contrast_enhancer.enhance(contrast)
    
    # Apply saturation adjustment
    color_enhancer = ImageEnhance.Color(adjusted_image)
    current_adjusted_image = color_enhancer.enhance(saturation)
    
    # Display
    plt.figure(figsize=(12, 8))
    plt.imshow(np.array(current_adjusted_image))
    plt.axis('off')
    plt.title(f'Exposure: {exposure:.2f} | Gamma: {gamma:.2f} | Contrast: {contrast:.2f} | Saturation: {saturation:.2f}', 
              fontsize=12)
    plt.tight_layout()
    plt.show()

def save_image(b):
    """Save the currently adjusted image"""
    import os
    from datetime import datetime
    
    if current_adjusted_image is None:
        save_status.value = "⚠ No image to save. Adjust settings first."
        return
    
    # Create save directory if it doesn't exist
    save_dir = "test\\images"
    os.makedirs(save_dir, exist_ok=True)
    
    # Generate filename with timestamp and settings
    timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
    exp_val = exposure_slider.value
    gamma_val = gamma_slider.value
    cont_val = contrast_slider.value
    sat_val = saturation_slider.value
    
    filename = f"adjusted_image_E{exp_val:.1f}_G{gamma_val:.1f}_C{cont_val:.1f}_S{sat_val:.1f}_{timestamp}.jpg"
    filepath = os.path.join(save_dir, filename)
    
    # Save the image
    current_adjusted_image.save(filepath, quality=95)
    
    # Update save status
    save_status.value = f"✓ Saved: {filename}"
    print(f"Image saved to: {filepath}")

# Create control widgets
exposure_slider = widgets.FloatSlider(
    value=1.0, min=0.1, max=3.0, step=0.1,
    description='Exposure:', continuous_update=False,
    style={'description_width': '80px'},
    layout=widgets.Layout(width='400px')
)

gamma_slider = widgets.FloatSlider(
    value=1.0, min=0.5, max=2.5, step=0.1,
    description='Gamma:', continuous_update=False,
    style={'description_width': '80px'},
    layout=widgets.Layout(width='400px')
)

contrast_slider = widgets.FloatSlider(
    value=1.0, min=0.5, max=2.0, step=0.1,
    description='Contrast:', continuous_update=False,
    style={'description_width': '80px'},
    layout=widgets.Layout(width='400px')
)

saturation_slider = widgets.FloatSlider(
    value=1.0, min=0.0, max=2.0, step=0.1,
    description='Saturation:', continuous_update=False,
    style={'description_width': '80px'},
    layout=widgets.Layout(width='400px')
)

# Reset button
def reset_values(b):
    exposure_slider.value = 1.0
    gamma_slider.value = 1.0
    contrast_slider.value = 1.0
    saturation_slider.value = 1.0

reset_button = widgets.Button(
    description='Reset to Original',
    button_style='info',
    layout=widgets.Layout(width='150px')
)
reset_button.on_click(reset_values)

# Create the interactive interface
print("Advanced Image Exposure & Color Controls")
print("=======================================")
print("• Exposure: Overall brightness multiplier")
print("• Gamma: Midtone adjustment (>1 = brighter mids, <1 = darker mids)")
print("• Contrast: Contrast enhancement")
print("• Saturation: Color intensity")
print()

# Create save button
save_button = widgets.Button(
    description='Save Image',
    button_style='success',
    icon='download',
    layout=widgets.Layout(width='120px')
)
save_button.on_click(save_image)

# Create save status label
save_status = widgets.Label(
    value="Ready to save...",
    layout=widgets.Layout(width='350px')
)

# Layout widgets
controls = widgets.VBox([
    exposure_slider,
    gamma_slider,
    contrast_slider,
    saturation_slider,
    widgets.HBox([reset_button, save_button]),
    save_status
])

# Create interactive widget
interactive_advanced = widgets.interact(
    update_advanced_image,
    exposure=exposure_slider,
    gamma=gamma_slider,
    contrast=contrast_slider,
    saturation=saturation_slider
)

display(controls)
```
